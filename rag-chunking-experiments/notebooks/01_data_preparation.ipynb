# 01. 데이터 준비

RAG 청킹 실험을 위한 데이터 수집 및 전처리

## 1. 환경 설정

```python
import sys
sys.path.append('..')

import json
from pathlib import Path
from src.utils.document_loader import DocumentLoader
```

## 2. 원본 문서 로드

```python
# 문서 로더 초기화
loader = DocumentLoader(data_dir='../data/raw')

# 모든 문서 로드
documents = loader.load_all(extensions=['.txt', '.md'])

print(f"총 {len(documents)}개 문서 로드됨")
```

## 3. 문서 통계 확인

```python
# 문서별 통계
print("=" * 60)
print("문서별 통계")
print("=" * 60)

for doc in documents:
    text = doc['text']
    word_count = len(text.split())
    char_count = len(text)
    line_count = len(text.split('\n'))
    
    print(f"\n[{doc['id']}]")
    print(f"  - 단어 수: {word_count:,}")
    print(f"  - 문자 수: {char_count:,}")
    print(f"  - 줄 수: {line_count:,}")
```

```python
# 전체 통계
stats = loader.get_statistics(documents)

print("\n" + "=" * 60)
print("전체 통계")
print("=" * 60)
for key, value in stats.items():
    if isinstance(value, float):
        print(f"{key}: {value:,.1f}")
    else:
        print(f"{key}: {value:,}")
```

## 4. 문서 내용 미리보기

```python
# 첫 번째 문서 미리보기
if documents:
    doc = documents[0]
    print(f"문서: {doc['id']}")
    print("=" * 60)
    print(doc['text'][:2000] + "...")
```

## 5. 전처리된 문서 저장

```python
# 저장 경로 생성
output_path = Path('../data/processed')
output_path.mkdir(parents=True, exist_ok=True)

# 전처리된 문서 저장
loader.save_processed(documents, '../data/processed/cleaned_documents.json')
```

## 6. 테스트 쿼리 확인

```python
# 테스트 쿼리 로드
test_queries = loader.load_test_queries('../data/evaluation/test_queries.json')

print(f"\n총 {len(test_queries)}개 테스트 쿼리 로드됨")
```

```python
# 쿼리 카테고리 및 난이도 분포
from collections import Counter

categories = Counter([q['category'] for q in test_queries])
difficulties = Counter([q['difficulty'] for q in test_queries])

print("\n카테고리 분포:")
for cat, count in categories.items():
    print(f"  - {cat}: {count}개")

print("\n난이도 분포:")
for diff, count in difficulties.items():
    print(f"  - {diff}: {count}개")
```

```python
# 샘플 쿼리 확인
print("\n샘플 쿼리:")
print("=" * 60)
for query in test_queries[:5]:
    print(f"[{query['query_id']}] ({query['category']}/{query['difficulty']})")
    print(f"  Q: {query['query']}")
    print()
```
